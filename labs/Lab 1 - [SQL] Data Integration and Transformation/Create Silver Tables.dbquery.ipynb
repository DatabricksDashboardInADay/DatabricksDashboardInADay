```sql

-- Date Silver Table
CREATE OR REPLACE TABLE  silver.s_date AS
SELECT * FROM sunny_bay_roastery.bronze.date;

-- Store Silver Table
CREATE OR REPLACE TABLE  silver.s_store AS
SELECT * FROM sunny_bay_roastery.bronze.store;

-- Customer Silver Table
CREATE OR REPLACE TABLE  silver.s_customer AS
SELECT * FROM sunny_bay_roastery.bronze.customer;

-- Product  Silver Table
CREATE OR REPLACE TABLE  silver.s_product AS
SELECT * FROM sunny_bay_roastery.bronze.product;

-- Coffee sales Silver Table
CREATE OR REPLACE TABLE  silver.s_sales AS
SELECT * FROM sunny_bay_roastery.bronze.sales WHERE quantity_sold > 0;

```
