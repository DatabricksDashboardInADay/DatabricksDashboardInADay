````python
dbutils.widgets.text("catalog", "sunny_bay_roastery")
dbutils.widgets.text("bronze_schema", "bronze")
dbutils.widgets.text("silver_schema", "silver")
dbutils.widgets.text("gold_schema", "gold")
dbutils.widgets.text("volume", "raw")
````

````python
# Validate that bronze, silver, and gold schemas are not all the same
bronze = dbutils.widgets.get("bronze_schema")
silver = dbutils.widgets.get("silver_schema")
gold = dbutils.widgets.get("gold_schema")

if bronze == silver or silver == gold or bronze == gold:
    raise ValueError(
        f"❌ Error: Two or more schemas share the same name. This may cause conflicts. "
        "This will cause table-name collisions in the pipeline. "
        "Please use different schema names for bronze, silver, and gold layers."
    )

else:
    print(f"✅ Schema validation passed: Bronze='{bronze}', Silver='{silver}', Gold='{gold}'")
````
